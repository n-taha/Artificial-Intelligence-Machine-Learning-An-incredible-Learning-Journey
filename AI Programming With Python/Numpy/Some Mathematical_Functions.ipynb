```python
#some mathematical Function of Numpy
import numpy as np

arr = [1,2,3,4,5]
print(np.sum(arr))
print(np.mean(arr))
print(np.std(arr))
print(np.min(arr))
print(np.max(arr))
```

```python
#trigonometric function
arr = np.array([np.pi/2, np.pi])
arr2 = np.array([90, 180, 120])
arr_3 = np.deg2rad(arr2)
print(np.cos(arr_3))
print(np.sin(arr))
print(np.tan(arr_3))
```

```python
# some mathematical functions

arr = np.array([1,2,3,4])

print(np.exp(arr))
print(np.log10(arr))
print(np.log(arr)) #it means ln
```

```python
arr = np.array([1.2, 2.6, 3.9])
print(np.round(arr)) # round to nearest
print(np.floor(arr)) # round down
print(np.ceil(arr))# round up

powers = np.array([1,2,3,4])
print(np.power(powers, 2))
```
