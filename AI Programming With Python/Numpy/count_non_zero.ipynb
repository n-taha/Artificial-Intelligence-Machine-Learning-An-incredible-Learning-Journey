```python
import numpy as np

arr = np.array([1,2,3,4,0])
count = np.count_nonzero(arr)
print(count)
```

```python
#boolean count

arr = np.array([1,2,3,4,5])
print(np.count_nonzero(arr > 3))
```

```python
# unique value count

arr = np.array([1,2,1,3,4,3,2,3,4,1])
unique = np.unique(arr)
print(unique)
```

```python
# return count
arr = np.array([1,2,1,3,4,3,2,3,4,1])
unique, return_c = np.unique(arr, return_counts=True)
print(unique)
print(return_c) # which value how many times come
```

```python
arr = np.array(["apple", "mango", "banana", "apple", "mango"])
unique , count = np.unique(arr, return_counts=True)
print(unique)
print(count)
frq = dict(zip(unique, count))
print(frq)
```

```python
#multidimentional Counting

arr = np.array([
    [1,0,3],
    [4,5,0]
])

print(np.count_nonzero(arr, axis=0))
print(np.count_nonzero(arr, axis=1))
```

```python
# counting Nan

arr = np.array([1,2,3,4,np.nan, np.nan])
print(np.count_nonzero(np.isnan(arr)))
print(np.isnan(arr))
```

```python
arr = np.array([12, 15, 18, 10, 25])
mask = arr > 15
print(mask)
print(np.sum(mask))
```
