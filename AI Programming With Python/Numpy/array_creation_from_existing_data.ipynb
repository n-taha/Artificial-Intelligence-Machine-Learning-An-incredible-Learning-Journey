```python
import numpy as np
num = [1,2,3,4.5]

arr = np.array(num, dtype=np.int32)
print(arr)
print(arr.dtype)
```

```python
#mixed List

mxd = [1, '3', True]
arr = np.array(mxd)

print(arr)
print(arr.dtype)
```

```python
matrix = [
    [1,2,3,4],
    [5,6,7,8]
]

arr = np.array(matrix)
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
tpl = (1,2,3,4)
arr = np.array(tpl, dtype=np.int16)

print(arr)
print(arr.dtype)

arr = arr.astype(np.int32)
print(arr)
print(arr.dtype)
```

```python
sts = {1,2,3,4}
arr = np.array(list(sts))
print(arr)
print(arr.dtype)
```

```python
# dictionaries
dicts = {
    "name": "Taha",
    "roll": 1460,
    "form": "A"
}

arr = np.array(list(dicts.keys()))
arr2 = np.array(list(dicts.values()))
arr3 = np.array(list(dicts.items()))
print(arr)
print(arr2)
print(arr3)
print(arr3.shape)
print(arr3.ndim)
```
