```python
# Advanced Indexing
import numpy as np

arr = np.array([1,2,3,4,5])
values = arr[[0,2,3]]
print(values)
```

```python
#2d Array Indexing
arr2 = np.array([
    [1,2,3],
    [4,5,6]
])

values2 = arr2[[0, 1], [1, 2]]
print(values2)
```

```python
# boolen Indexing
arr3 = np.array([
    [1,2,3],
    [4,5,6]
])

values3 = arr3[arr3 > 1]
print(values3)

```

```python
# 3d array slicing
arr4 = np.array([
    [
        [1,2,3],
        [4,5,6]
    ],
    [
        [4,5,6],
        [7,8,9]
    ]
])

print(arr4.ndim)

for i in np.nditer(arr4):
    print(i)
```
