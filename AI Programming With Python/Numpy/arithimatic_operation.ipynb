```python
import numpy as np

arr = np.array([1,2,3,4,5])
arr_2 = np.array([10,20,30,40,50])

print(arr + arr_2)
print(arr - arr_2)
print(arr * arr_2)
print(arr / arr_2)
```

```python
#scaler Operation
arr = np.array([1,2,3,4,5])
print(arr + 10) 
print(arr - 10) 
print(arr * 10) 
print(arr / 10) 
```

```python
arr = np.array([1,2,3,4,5]) # False Of Array

print(arr > 100)
print(arr > 2)
```

```python
# in-place Operation Of Array
arr = np.array([1,2,3,4])

arr+=5;
print(arr)
```
