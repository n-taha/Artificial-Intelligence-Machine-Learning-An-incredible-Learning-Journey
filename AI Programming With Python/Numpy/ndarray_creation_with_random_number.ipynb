```python
import numpy as np

arr = np.random.rand(2,3) # 2 row, 3 column -> value 0 to under 1
print(arr)
```

```python
arr = np.random.randint(1, 10, (3,3)) # value will be 1 to 9 and shape -> 3 x 3
print(arr)
```

```python
arr = np.random.uniform(50, 100, (3, 3, 3)) # Just Carry Float Value
print(arr)
```

```python
arr = np.arange(1, 10, 1)
print(arr)
mat = arr.reshape(3,3)
print(mat)
```

```python
# Numpy Line Space

line = np.linspace(0, 4, 6)
print(line)
```

```python
#numpy logspcae
log = np.logspace(0, 4, 6, base=2)
print(log)
```

```python
log = np.logspace(0, 4, 6)
print(log)
```
