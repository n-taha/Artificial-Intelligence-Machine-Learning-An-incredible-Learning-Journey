```python
import numpy as np
arr = np.array([1,2,3])
print(arr.dtype)
arr = np.array([1.2, 2.2, 3.4])
print(arr.dtype)
```

```python
#upcasting to float
arr = np.array([1,3,5.6])
print(arr.dtype)
```

```python
#upcasting to string
arr = np.array([1,2,3.4, "hello"])
print(arr.dtype) # here U32 means String and U = Unicode String
```

```python
arr = np.array([1,2,3.5, True, "h"])
print(arr.dtype)
```

```python
arr = np.array([1,2,3,4], dtype=np.int16) # we change the data type
print(arr.dtype)
```

```python
#changing data type using astype
arr = np.array([1,2,3,4,5.3]) # float64
print(arr.dtype)
arr2 = arr.astype(dtype=np.int32) # [1,2,3,4,5]  The clone of arr removed decimal number
print(arr2)
print(arr2.dtype)
print(arr.dtype)
```
