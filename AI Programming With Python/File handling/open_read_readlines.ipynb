```python
file = open("sample.txt", "r");
content = file.readlines()
print(content);  # List Of EveryLine 
```

```python
for line in content:
    print(line, end="") #we can print this list using this kind of loop
```

```python
file.close()
print(file.closed) # Must We have to close the file, what we opened for work. if we don't close the file. there are a big risk of data leackage
```

```python
# if we follow this approch, we don't need to close file manually
with open("sample.txt", "r") as file:
    content = file.readlines()
    for line in content:
        print(line, end="");
    # in with block file is open
print(file.closed) # outside this block file is close.
```
