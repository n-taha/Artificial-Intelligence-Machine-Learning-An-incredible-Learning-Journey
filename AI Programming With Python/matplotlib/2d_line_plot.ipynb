```python
hours = [2,3,5,6,3,4]
days = [1,2,3,4,5,6]
plt.plot(days, hours)
plt.xlabel("Days")
plt.ylabel("Hours")
plt.title("Treands Of Study Hour Per Day")
```

```python
df = pd.read_csv("enrollment_data.csv")
df
```

```python
plt.xlabel("Programming Enrollment")
plt.ylabel("Years")
plt.title("Programming Student Enrollment Trend Per Year")

plt.plot(df["Year"], df["Programming"])
```

```python
plt.ylabel("Programming & Digital Marketing Enrollment")
plt.xlabel("Years")
plt.title("Programming & Digital Marketing Student Enrollment Trend Per Year")

plt.plot(df["Year"], df["Programming"])
plt.plot(df["Year"], df["Digital Marketing"])
```

```python
plt.ylabel("Programming & Digital Marketing Enrollment")
plt.xlabel("Years")
plt.title("Programming & Digital Marketing Student Enrollment Trend Per Year")

plt.plot(df["Year"], df["Programming"], label="Programming", color="green", linewidth=2, linestyle="dashed", marker='o', markersize=4)
plt.plot(df["Year"], df["Digital Marketing"], label="Digital Markeing", color="orange", linewidth=3, linestyle="dotted", marker="x", markersize=4)
plt.grid()
plt.legend()
```
