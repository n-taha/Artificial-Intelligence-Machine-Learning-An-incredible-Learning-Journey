```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("student_IQdata.csv")
df
```

```python
# Relation Between Study Hour And IQ Score -> Positive Correlation
plt.xlabel("Study Hour")
plt.ylabel("IQ Score")
plt.title("Scatter Plot Of Student Study Hour and IQ Score")

plt.scatter(df["Study_Hour"], df["IQ_Score"], marker='o', color="red")
```

```python
# Relation Between Study Hour And Shoe Size -> No Correlation
plt.xlabel("shoe size")
plt.ylabel("IQ Score")
plt.title("Scatter Plot Of Shoe Size and IQ Score")

plt.scatter(df["Shoe_Size"], df["IQ_Score"])
```

```python
# Relation Between Chilling Time And IQ Score -> Negative Correlation

plt.xlabel("Chilling Time")
plt.ylabel("IQ Score")
plt.title("Scatter Plot Of Student Chilling Time & IQ Score")

plt.scatter(df["Chilling_Hours"], df["IQ_Score"], marker='o', color="red")
```
