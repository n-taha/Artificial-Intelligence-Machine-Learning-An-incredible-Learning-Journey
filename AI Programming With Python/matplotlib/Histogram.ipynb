```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
students = pd.read_csv("student_IQdata.csv")
students
```

```python
plt.hist(students["IQ_Score"], color="red", edgecolor="black")
plt.xlabel("IQ Score")
plt.ylabel("Frequency Of Students")
```

```python
plt.hist(students["Study_Hour"] , color="green", edgecolor="orange")
plt.xlabel("Study Hour")
plt.ylabel("Frequency Of Student")
```

```python
plt.hist(students["Study_Hour"] , bins =[1,2,3,4,5,6],color="green", edgecolor="orange")
plt.xlabel("Study Hour")
plt.ylabel("Frequency Of Student")
```
