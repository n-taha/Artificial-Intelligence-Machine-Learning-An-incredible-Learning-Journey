```python
import pandas as pd 
my_list = [["Alice", 20], ["Bob", 30]]
df = pd.DataFrame(my_list, columns=["Name", "Age"], index=[1,2])
df
```

```python
my_tup = (("Alice", 40), ("Bob", 40))
df = pd.DataFrame(my_tup, index=[1,2], columns=["name", "age"])
df
```
