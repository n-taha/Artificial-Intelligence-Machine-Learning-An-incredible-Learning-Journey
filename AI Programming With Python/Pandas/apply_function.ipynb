```python
import pandas as pd

df = pd.read_csv("student_data.csv")
df
```

```python
df["Total Marks"] = df["Data Structure Marks"] + df["Python Marks"]
```

```python
mn = df["Total Marks"].min()
mx = df["Total Marks"].max()
df["Scaled Marks"] = df["Total Marks"].apply(lambda x : (x - mn)/(mx - mn))
df
```

```python
def grading_system(marks):
    if marks >= 160:
        return "A+"
    elif marks >= 140:
        return "A"
    else:
        return "Nothing"

df["Grade"] = df["Total Marks"].apply(grading_system)
df
```

```python
def demo_func(df):
    a = df["Data Structure Marks"] * 2
    b = df["Python Marks"] * 2
    c = df["Algorithm Marks"] * 2 

    return a + b + c 

df["Exception Marks"] = df.apply(demo_func, axis=1)
df 
```
